# Visualisations 
**Data:** `data/features_and_embeddings.csv`

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from pathlib import Path

from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler

DATA = Path("../data")
final = pd.read_csv(DATA / "features_and_embeddings.csv")

feature_cols = [c for c in final.columns if c not in ("transcript", "theme", "y")]
X = final[feature_cols]
y = final["y"]

print("Shape:", final.shape)
print("Features:", len(feature_cols))
print(final["y"].map({0: "human", 1: "AI"}).value_counts())
```

```python
# See the correlation of each feature and the target
corr = X.corrwith(y, method="spearman").sort_values()
corr.plot.barh(figsize=(6, 10), color=["#d9795b" if v < 0 else "#4f9d8f" for v in corr])
plt.title("Correlation with y (right = AI, left = human)")
plt.show()
```

```python
# Correlation HeatMap
corr_matrix = X.corr(method="spearman")

plt.figure(figsize=(14, 12))
sns.heatmap(corr_matrix, cmap="vlag", center=0, vmin=-1, vmax=1, square=True,
            xticklabels=True, yticklabels=True, cbar_kws={"shrink": 0.6})
plt.title("Feature correlations (Spearman)")
plt.tight_layout()
plt.show()
```

```python
# Distributions of the top features
top = corr.abs().sort_values(ascending=False).head(8).index
fig, axes = plt.subplots(2, 4, figsize=(16, 7))
for ax, c in zip(axes.ravel(), top):
    sns.boxplot(data=final, x=final["y"].map({0: "human", 1: "AI"}), y=c, ax=ax)
    ax.set_xlabel("")
plt.tight_layout()
plt.show()
```

```python
# See which feature the model leans on
model = LogisticRegression(max_iter=2000).fit(StandardScaler().fit_transform(X), y)
coef = pd.Series(model.coef_[0], index=feature_cols)
top12 = coef.reindex(coef.abs().sort_values(ascending=False).head(12).index).sort_values()

top12.plot.barh(figsize=(8, 5), color=["#d9795b" if v < 0 else "#4f9d8f" for v in top12])
plt.title("Strongest model associations (right = AI, left = human)")
plt.show()
```

```python
# Compare themes
print(pd.crosstab(final["theme"], final["y"].map({0: "human", 1: "AI"})))

for t, d in final.groupby("theme"):
    m = LogisticRegression(max_iter=2000).fit(StandardScaler().fit_transform(d[feature_cols]), d["y"])
    c = pd.Series(m.coef_[0], index=feature_cols)
    c = c.reindex(c.abs().sort_values(ascending=False).head(8).index).sort_values()
    c.plot.barh(figsize=(7, 4), color=["#d9795b" if v < 0 else "#4f9d8f" for v in c])
    plt.title(f"Theme: {t}")
    plt.show()
```
