```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load in the data_temp.csv
data_temp = pd.read_csv('features.csv')

# All features and their definitions
feature_definitions = {
    "n_words_calc":         "Total number of words in the transcript (recalculated).",
    "lex_var_30":           "Average number of unique words in rolling windows of 30 words (local vocabulary diversity).",
    "mattr_30":             "Moving Average Type-Token Ratio over windows of 30 words (vocabulary richness controlling for length).",
    "negation_rate":        "Frequency of negation words (not, no, never, don't, etc.) per 100 words.",
    "reaction_rate":        "Frequency of reaction/interjection words (oh, wow, ugh, whoa, etc.) per 100 words.",
    "contraction_rate":     "Percentage of words that are contractions (don't, it's, I'm, etc.).",
    "everything_count":     "Absolute count of the word 'everything'.",
    "love_count":           "Absolute count of the word 'love'.",
    "well_count":           "Absolute count of the word 'well'.",
    "filler_rate":          "Density of conversational filler words (um, uh, like, you know, basically, etc.) per 100 words.",
    "everything_rate":      "Rate of the word 'everything' per 100 words.",
    "love_rate":            "Rate of the word 'love' per 100 words.",
    "well_rate":            "Rate of the word 'well' per 100 words.",
    "n_sentences":          "Total number of sentences in the transcript.",
    "avg_sent_len":         "Average number of words per sentence.",
    "sent_len_std":         "Standard deviation of sentence lengths (stylistic variety).",
    "max_sent_len":         "Length (in words) of the longest sentence.",
    "short_sent_rate":      "Percentage of sentences that are very short (≤ 5 words).",
    "question_rate":        "Percentage of sentences that end with a question mark.",
    "exclaim_rate":         "Percentage of sentences that end with an exclamation mark.",
    "ellipsis_rate":        "Frequency of ellipses ('...') relative to number of sentences.",
    "avg_word_len":         "Average number of characters per word.",
    "long_word_rate":       "Percentage of words that are long (≥ 7 characters).",
    "hapax_rate":           "Percentage of unique words that appear only once in the text.",
    "ttr_full":             "Type-Token Ratio of the entire transcript (overall lexical diversity).",
    "top_word_share":       "Percentage of the text made up by the single most frequent word.",
    "profanity_rate":       "Frequency of profane/vulgar words per 100 words.",
    "digit_rate":           "Frequency of pure digit tokens (e.g. '3', '45') per 100 words.",
    "first_person_rate":    "Frequency of first-person pronouns (I, me, my, we, us, our...) per 100 words.",
    "second_person_rate":   "Frequency of second-person pronouns (you, your, yours...) per 100 words.",
    "we_rate":              "Frequency of collective first-person pronouns (we, us, our...) per 100 words.",
    "repeat_bigram_rate":   "Rate of repeated adjacent word pairs (bigrams).",
    "repeat_sentence_rate": "Rate of identical sentences that appear more than once.",
    "proper_noun_rate":     "Heuristic density of capitalized proper nouns (names, places, brands).",
    "cta_flag":             "Binary flag (1/0) indicating presence of a Call-To-Action (subscribe, like, comment, follow, etc.)."
}
```

```python
# Only keep numeric columns
numeric_df = data_temp.select_dtypes(include='number')

# Heatmap of correlations with the target 'y'
plt.figure(figsize=(12, 10))
sns.heatmap(
    numeric_df.corr()[['y']].sort_values(by='y', ascending=False),
    annot=True,
    cmap='coolwarm',
    center=0
)
plt.title("Feature Correlations with Target (y)")
plt.tight_layout()
plt.show()
 
# Features with correlation greater than 0.1 with the target 'y'
important_features = [
    'filler_rate', 'avg_word_len', 'long_word_rate', 'repeat_bigram_rate',
    'mattr_30', 'lex_var_30', 'everything_count', 'hapax_rate',
    'reaction_rate', 'well_rate', 'well_count', 'ttr_full'
]

for feature in important_features:
    plt.figure(figsize=(8, 4))
    sns.histplot(data=data_temp, x=feature, hue='y', kde=True, element="step", common_norm=False)
    plt.title(f"Distribution of {feature} – AI vs Human")
    plt.show()

for feature in important_features:
    plt.figure(figsize=(7, 4))
    sns.boxplot(data=data_temp, x='y', y=feature)
    plt.title(f"{feature} – AI vs Human")
    plt.show()
```
